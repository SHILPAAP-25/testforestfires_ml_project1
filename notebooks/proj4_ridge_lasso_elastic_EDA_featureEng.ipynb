ALGERIAN FOREST FIRES DATASET 

Data Set Information :

The datset includes 224 instances taht regroup a data of two regions of Algeria , namely the Bejajia region located in the northeast of Algeria and the Sidi Bel-abbes region located in the northwest of Algeria. 

122 instances for each region 

the period from june 2012 to septem,ber 2012. The data set includes 11  attributes and 1 output attribute (class). the 244 instances have been classified into fire( 138 classes) and not fir( 106 classes ) classes .

DATA CLEANING

```python
##missing values 
dataset.isnull().any(axis=1)
```

```python
##missing values 

dataset[dataset.isnull().any(axis=1)]
```

the datset is converted into two sets based on region from 122th index, we can make a new column based on the region. 

1. "Bejaia Region Dataset"

2. "Sido-bel abbes region dataset"

add new column with region 

```python
dataset.loc[:122,"Region"]=0
dataset.loc[122:,"Region"]=1

dff=dataset
```

```python
dff.info()
```

```python
dff[['Region']]=dff[['Region']].astype(int)
```

```python
dff.head()
```

```python
dff.isnull().sum()
```

```python
## remove the null avlues 

df=dff.dropna().reset_index(drop=True)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.iloc[[122]]
```

```python
##remove the 122nd row

df=df.drop(122).reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
df.columns
```

```python
##fix spaces in columns names

df.columns=df.columns.str.strip()

df.columns
```

```python
df.info()
```

##changes  the required columns as integer data type

```python
df.columns
```

```python
dff[['months','day','year','Temperature','RH','Ws']]=df[['month','day','year','Temperature','RH','Ws']].astype(int)
```

```python
df.info()
```

###cagngeing the other columns to float data type 

```python
objects=[features for features in df.columns if df[features].dtypes=='O']
```

```python
for i in objects:
    if i!= 'Classes':
        df[i]=df[i].astype(float)
```
